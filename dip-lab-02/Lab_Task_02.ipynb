```python
import cv2 as cv
import matplotlib.pyplot as plt
import numpy as np
print("Libraries imported successfully")
```

```python
IMAGE_PATH="my_photo.jpg"
image_bgr = cv.imread(IMAGE_PATH, cv.IMREAD_COLOR)
print("Loaded:", image_bgr is not None)
print("Shape:", image_bgr.shape)
```

```python
plt.figure()
plt.imshow(cv.cvtColor(image_bgr, cv.COLOR_BGR2RGB))
plt.title("My Photo")
plt.axis('off')
plt.show()
```

```python
# BGR to GRAY
image_gray = cv.cvtColor(image_bgr, cv.COLOR_BGR2GRAY)

plt.figure(figsize=(10,5))
plt.subplot(1,2,1)
plt.imshow(cv.cvtColor(image_bgr, cv.COLOR_BGR2RGB))
plt.title("Original BGR")
plt.axis('off')

plt.subplot(1,2,2)
plt.imshow(image_gray, cmap='gray')
plt.title("Grayscale")
plt.axis('off')
plt.show()

print("Gray Shape:", image_gray.shape)
```

```python

resized = cv.resize(image_bgr, (400, 400))

plt.imshow(cv.cvtColor(resized, cv.COLOR_BGR2RGB))
plt.title("Resized 400x400")
plt.axis('off')
plt.show()
```

```python
cv.imwrite("my_photo_gray.jpg", image_gray)
print("Gray image saved!")
```

```python
b, g, r = cv.split(image_bgr)

plt.figure(figsize=(12,4))
plt.subplot(1,3,1)
plt.imshow(b, cmap='Blues')
plt.title("Blue Channel")
plt.axis('off')

plt.subplot(1,3,2)
plt.imshow(g, cmap='Greens')
plt.title("Green Channel")
plt.axis('off')

plt.subplot(1,3,3)
plt.imshow(r, cmap='Reds')
plt.title("Red Channel")
plt.axis('off')
plt.show()
```

```python
flip_horizontal = cv.flip(image_bgr, 1) # 1 = horizontal
flip_vertical = cv.flip(image_bgr, 0)   # 0 = vertical

plt.figure(figsize=(10,4))
plt.subplot(1,2,1)
plt.imshow(cv.cvtColor(flip_horizontal, cv.COLOR_BGR2RGB))
plt.title("Horizontal Flip")
plt.axis('off')

plt.subplot(1,2,2)
plt.imshow(cv.cvtColor(flip_vertical, cv.COLOR_BGR2RGB))
plt.title("Vertical Flip")
plt.axis('off')
plt.show()
```

```python
plt.figure()
plt.hist(image_gray.ravel(), bins=256, range=[0,256])
plt.title("Grayscale Histogram")
plt.xlabel("Pixel Value")
plt.ylabel("Frequency")
plt.show()
```

```python
# Gaussian Blur
blur = cv.GaussianBlur(image_bgr, (15, 15), 0)

plt.figure(figsize=(10,4))
plt.subplot(1,2,1)
plt.imshow(cv.cvtColor(image_bgr, cv.COLOR_BGR2RGB))
plt.title("Original")
plt.axis('off')

plt.subplot(1,2,2)
plt.imshow(cv.cvtColor(blur, cv.COLOR_BGR2RGB))
plt.title("Gaussian Blur (15,15)")
plt.axis('off')
plt.show()
```

```python
# Threshold lagao
_, thresh = cv.threshold(image_gray, 120, 255, cv.THRESH_BINARY)

plt.figure(figsize=(10,4))
plt.subplot(1,2,1)
plt.imshow(image_gray, cmap='gray')
plt.title("Grayscale")
plt.axis('off')

plt.subplot(1,2,2)
plt.imshow(thresh, cmap='gray')
plt.title("Binary Threshold (120)")
plt.axis('off')
plt.show()

print("Thresholding Done!")
```

```python
cv.imwrite("output_blur.jpg", blur)
cv.imwrite("output_thresh.jpg", thresh)
print("All outputs saved in dip-lab-02 folder!")
```
